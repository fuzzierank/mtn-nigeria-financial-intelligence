# MTN Nigeria Financial Intelligence Analysis (2021–2025)

First reproducible analysis of the working dataset.

```
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/mtn_nigeria_financial_intelligence_2021_2025.csv")
df.head()
```

## Revenue and data transformation

```
df["data_revenue_share_pct"] = df["data_revenue_ngn_bn"] / df["service_revenue_ngn_bn"] * 100
df[["year", "total_revenue_ngn_bn", "data_revenue_ngn_bn", "data_revenue_share_pct"]]
```

```
plt.figure(figsize=(9,5))
plt.plot(df["year"], df["total_revenue_ngn_bn"], marker="o", label="Total revenue")
plt.plot(df["year"], df["data_revenue_ngn_bn"], marker="o", label="Data revenue")
plt.title("MTN Nigeria: Total Revenue vs Data Revenue")
plt.xlabel("Year")
plt.ylabel("₦ billion")
plt.grid(alpha=.25)
plt.legend()
plt.show()
```

## Profitability

```
df[["year", "ebitda_ngn_bn", "pat_ngn_bn", "free_cash_flow_ngn_bn"]]
```

```
plt.figure(figsize=(9,5))
plt.plot(df["year"], df["ebitda_ngn_bn"], marker="o", label="EBITDA")
plt.plot(df["year"], df["pat_ngn_bn"], marker="o", label="Profit after tax")
plt.axhline(0, linewidth=.8)
plt.title("MTN Nigeria: EBITDA vs Profit After Tax")
plt.xlabel("Year")
plt.ylabel("₦ billion")
plt.grid(alpha=.25)
plt.legend()
plt.show()
```

## Data consumption

```
df[["year", "data_subscribers_m", "data_usage_gb_per_active_user"]]
```

```
fig, ax1 = plt.subplots(figsize=(9,5))
ax1.plot(df["year"], df["data_subscribers_m"], marker="o")
ax1.set_xlabel("Year")
ax1.set_ylabel("Data subscribers (million)")
ax1.grid(alpha=.25)
ax2 = ax1.twinx()
ax2.plot(df["year"], df["data_usage_gb_per_active_user"], marker="s")
ax2.set_ylabel("GB per active user")
plt.title("MTN Nigeria: Data Subscribers and Usage")
plt.show()
```
